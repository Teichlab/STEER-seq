```r
markersPeaks <- readRDS('/lustre/scratch125/casm/teams/team215/users/od5/projects/celltherapy/ipsc/multiome/archr/tf_pilot/batch1/ipsc/archr_DARs_perturbation_obj.rds')
```

```r
markersPeaks
```

```r
markerList <- getMarkers(markersPeaks, cutOff = "FDR <= 0.2")
```

```r
markerList

markerPeaks_df <- Reduce(bind_rows,lapply(1:length(markerList), function(i) {
  ## Cell type or state
  ct <- names(markerList)[i]
  ## Object of marker peaks
  gr <- markerList[[i]]
  if (nrow(gr) == 0) {
    return(NULL)
  } else {
     ## Getting peak names and stats
    peaks <- data.frame(chr=as.character(gr$seqnames), 
                        start=gr$start,
                        end=gr$end,
                        Log2FC=gr$Log2FC,
                        FDR=gr$FDR,
                        MeanDiff=gr$MeanDiff,
                        Cell_type=ct,
                        stringsAsFactors = F) %>%
      mutate(peak=paste0(chr, ':',start, '-',end))
    return(peaks)
  }
}))
```
