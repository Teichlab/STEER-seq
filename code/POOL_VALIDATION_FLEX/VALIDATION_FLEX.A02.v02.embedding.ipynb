# Celltypist

```python
import celltypist
from celltypist import models
```

```python
# Enabling `force_update = True` will overwrite existing (old) models.
models.download_models(force_update = True)
```

```python
predictions = celltypist.annotate(adata, model = 'Pan_Fetal_Human.pkl', majority_voting = True)
```

```python
# Get an `AnnData` with predicted labels embedded into the cell metadata columns.
adata = predictions.to_adata()
```

```python
neuron_prob = predictions.probability_matrix.loc[:,'NEURON']
adata.obs['neuron_prob'] = neuron_prob
```

```python
sc.pl.umap(adata, color=['majority_voting','sample'], ncols=1)
```

```python
adata.write_h5ad('write/validation_flex_celltypist.h5ad')
```
